# 03 — Statistical and Spatial Analysis

This notebook tests pricing differences and explores relationships among key numerical variables.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.statistical_analysis import room_type_anova, borough_price_ttest

DATA_PATH = PROJECT_ROOT / "data" / "processed" / "airbnb_nyc_cleaned.csv"
IMAGE_DIR = PROJECT_ROOT / "images"
IMAGE_DIR.mkdir(exist_ok=True)

df = pd.read_csv(DATA_PATH)
```

## Hypothesis 1: Prices differ across room types

**H₀:** Mean listing prices are equal across room types.  
**H₁:** At least one room type has a different mean price.

```python
anova_result = room_type_anova(df)
anova_result
```

```python
alpha = 0.05
if anova_result.p_value < alpha:
    print("Reject H₀: average prices differ significantly across room types.")
else:
    print("Fail to reject H₀: there is insufficient evidence of a difference.")
```

## Hypothesis 2: Manhattan and Brooklyn have different average prices

```python
ttest_result = borough_price_ttest(df, "Brooklyn", "Manhattan")
ttest_result
```

```python
if ttest_result.p_value < alpha:
    print("Reject H₀: Brooklyn and Manhattan have significantly different average prices.")
else:
    print("Fail to reject H₀: there is insufficient evidence of a difference.")
```

## Correlation analysis

```python
numeric_columns = [
    "price",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "calculated_host_listings_count",
    "availability_365"
]

correlation_matrix = df[numeric_columns].corr(numeric_only=True)
correlation_matrix
```

```python
plt.figure(figsize=(9, 7))
plt.imshow(correlation_matrix, aspect="auto")
plt.xticks(range(len(numeric_columns)), numeric_columns, rotation=45, ha="right")
plt.yticks(range(len(numeric_columns)), numeric_columns)
plt.colorbar(label="Correlation")
plt.title("Correlation Matrix")
plt.tight_layout()
plt.savefig(IMAGE_DIR / "correlation_matrix.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Business interpretation

Use statistical significance together with practical effect size and business context. A statistically significant difference does not automatically imply that every listing should follow the same pricing strategy.
